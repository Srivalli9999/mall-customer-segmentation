```python
# Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import LabelEncoder
```

```python
# Load Dataset
data = pd.read_csv("Mall_Customers.csv")
```

```python
#  Print First 20 Rows
print(data.head(20))
```

```python
# Convert Gender to Numbers
label = LabelEncoder()

data['Gender'] = label.fit_transform(data['Gender'])
```

```python
# Select Features for Clustering
X = data[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
# Apply K-Means
kmeans = KMeans( n_clusters=4,random_state=42 ,n_init=10)

data['Cluster'] = kmeans.fit_predict(X)
```

```python
# Print Cluster Results
print(data[['CustomerID', 'Cluster']].head())
```

```python
#Plotting

plt.figure(figsize=(10,7))
colors = ['red', 'blue', 'green', 'cyan', 'magenta']
labels = ['Cluster 0', 'Cluster 1', 'Cluster 2', 'Cluster 3', 'Cluster 4']

for i in range(5):
    plt.scatter(data[data['Cluster'] == i]['Annual Income (k$)'], 
                data[data['Cluster'] == i]['Spending Score (1-100)'], 
                s=100, c=colors[i], label=labels[i])

# Centroids - Yellow X
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], 
            s=300, c='yellow', marker='o', edgecolors='black', label='Centroids')

plt.title('Mall Customer Segments - 5 Clusters')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
cluster_sizes = data['Cluster'].value_counts().sort_index()
labels = [f'Cluster {i}' for i in range(5)]

plt.figure(figsize=(7,7))
plt.pie(cluster_sizes, labels=labels, autopct='%1.1f%%', startangle=90, 
        colors=['cornflowerblue', 'mediumorchid', 'plum', 'violet', 'orchid'])
plt.title('Cluster Size Distribution')
plt.axis('equal')
plt.show()
```
